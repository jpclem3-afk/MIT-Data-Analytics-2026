```python
dataframes = {
    "Customers": customers,
    "Products": products,
    "Sales": sales,
    "Stores": stores,
    "Inventory": inventory
}

for name, df in dataframes.items():
    print(f"\n{name} missing values:")
    print(df.isnull().sum())
```

```python
for name, df in dataframes.items():
    print(f"\n{name} data types:")
    print(df.dtypes)
```

```python
# Number of unique products
unique_products = products["product_id"].nunique()
print("Number of unique products:", unique_products)
```

```python
# Top 5 most expensive products
top_5_expensive = products.nlargest(
    5,
    "price"
)[["product_name", "price"]]

print("Top 5 most expensive products:")
display(top_5_expensive)
```

```python
# Store with the largest floor space
largest_store = stores.loc[
    stores["size_sqft"].idxmax(),
    ["store_name", "city", "state", "size_sqft"]
]

print("Store with the largest floor space:")
display(largest_store)
```

```python
# Distribution of customers by state
customer_state_distribution = customers["state"].value_counts()

print("Customer distribution by state:")
display(customer_state_distribution)
```

```python
# Convert date columns to datetime
for name, df in dataframes.items():
    date_columns = [
        column for column in df.columns
        if "date" in column.lower()
    ]

    for column in date_columns:
        df[column] = pd.to_datetime(df[column], errors="coerce")
        df[column] = df[column].ffill().bfill()
```

```python
for name, df in dataframes.items():

    # Replace missing numerical values with the column mean
    numerical_columns = df.select_dtypes(include="number").columns

    for column in numerical_columns:
        df[column] = df[column].fillna(df[column].mean())

    # Replace missing categorical values with the most frequent value
    categorical_columns = df.select_dtypes(
        include=["object", "category"]
    ).columns

    for column in categorical_columns:
        if not df[column].mode().empty:
            df[column] = df[column].fillna(df[column].mode()[0])
```

```python
for name, df in dataframes.items():
    print(f"\n{name} missing values after cleaning:")
    print(df.isnull().sum())
```

```python
print("Customer duplicates before cleaning:", customers.duplicated().sum())
print("Product duplicates before cleaning:", products.duplicated().sum())

customers = customers.drop_duplicates().reset_index(drop=True)
products = products.drop_duplicates().reset_index(drop=True)

print("Customer duplicates after cleaning:", customers.duplicated().sum())
print("Product duplicates after cleaning:", products.duplicated().sum())
```

## Discussion Questions

### 1. What are the key advantages of using Pandas for data cleaning?

Pandas makes it easy to inspect, filter, transform, and clean large datasets using concise Python code. It includes built-in methods for handling missing values, duplicates, incorrect data types, and summary statistics. The cleaning steps can also be repeated consistently.

### 2. How would handling missing values differ if the missing data had a pattern or meaning?

I would first investigate why the values were missing instead of automatically replacing them. If the missing data represented a meaningful customer behavior or business condition, I might create a separate missing-value category or indicator column to preserve that information.

### 3. What data-quality issues might not be visible through simple DataFrame inspection?

Potential issues include inconsistent spelling, incorrect values that still fall within a valid range, mismatched identifiers between datasets, inaccurate dates, formatting differences, hidden whitespace, and duplicate records that are not completely identical.

### 4. How would you document the cleaning process for reproducibility?

I would keep each cleaning step in an organized Jupyter notebook with Markdown explanations, comments, and clearly named variables. I would preserve the original data, record any assumptions, and save the cleaned results separately so another analyst could repeat the process.

### 5. When might it be better to remove rows with missing values?

Removing rows may be appropriate when only a small percentage of records are missing, when the missing values affect essential fields, or when imputing values could introduce bias or misleading information. It is most appropriate when removing the rows will not significantly reduce the dataset or distort the analysis.
