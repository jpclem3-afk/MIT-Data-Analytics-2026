```python
import pandas as pd

customers = pd.read_csv("customers_wk9_thurs.csv")

products = pd.read_csv("products_wk9_thurs.csv")

sales = pd.read_csv("sales_wk9_thurs.csv")
```

```python
print("Customers:")

display(customers.head())

print("Products:")

display(products.head())

print("Sales:")

display(sales.head())
```

```python
# Convert all date columns to DateTime format

customers["registration_date"] = pd.to_datetime(customers["registration_date"])

products["launch_date"] = pd.to_datetime(products["launch_date"])

sales["transaction_date"] = pd.to_datetime(sales["transaction_date"])
```

```python
# Extract useful date components from sales

sales["year"] = sales["transaction_date"].dt.year

sales["month"] = sales["transaction_date"].dt.month

sales["day_of_week"] = sales["transaction_date"].dt.day_name()

sales.head()
```

```python
# Display the sales date range

print("Earliest sales date:", sales["transaction_date"].min())

print("Latest sales date:", sales["transaction_date"].max())
```

```python
# Basic time analysis

print("Busiest day of the week:")

display(sales["day_of_week"].value_counts())

print("Month with the most transactions:")

display(sales["month"].value_counts())

print("Number of sales in each year:")

display(sales["year"].value_counts().sort_index())
```

```python
# Average product price by category

average_price_by_category = products.groupby("category")["price"].mean()

print("Average price by category:")

display(average_price_by_category)
```

```python
# Number of products in each category

product_count_by_category = products.groupby("category")["product_id"].count()

print("Number of products by category:")

display(product_count_by_category)
```

```python
# Average customer age in each state

average_age_by_state = customers.groupby("state")["age"].mean()

print("Average age by state:")

display(average_age_by_state)
```

```python
# Total spending for each customer

customer_spending = sales.groupby("customer_id")["total_amount"].sum()

customer_spending = customer_spending.sort_values(ascending=False)

print("Total spending per customer:")

display(customer_spending)
```

```python
# Customers who made the most purchases

purchase_counts = sales.groupby("customer_id")["sale_id"].count()

purchase_counts = purchase_counts.sort_values(ascending=False)

print("Customers with the most purchases:")

display(purchase_counts.head(10))
```

```python
# Total sales revenue by month

monthly_revenue = sales.groupby("month")["total_amount"].sum()

print("Total sales revenue by month:")

display(monthly_revenue)

```

```python
# Average transaction amount by day of the week

average_transaction_by_day = (

    sales.groupby("day_of_week")["total_amount"]

    .mean()

    .sort_values(ascending=False)

)

print("Average transaction amount by day:")

display(average_transaction_by_day)
```

```python
# Combine sales with product information

sales_products = sales.merge(

    products[["product_id", "category"]],

    on="product_id"

)

# Calculate total revenue by product category

category_revenue = (

    sales_products.groupby("category")["total_amount"]

    .sum()

    .sort_values(ascending=False)

)

print("Total revenue by product category:")

display(category_revenue)

print("\nHighest-revenue category:", category_revenue.idxmax())
```

```python
# Find the state with the most customers

customers_by_state = customers["state"].value_counts()

print("Number of customers by state:")

display(customers_by_state)

print("\nState with the most customers:", customers_by_state.idxmax())
```

```python
# Identify the top 5 customers by total spending

top_5_customers = (

    customer_spending.head(5)

    .reset_index()

    .merge(

        customers[["customer_id", "first_name", "last_name"]],

        on="customer_id"

    )

)

top_5_customers["full_name"] = (

    top_5_customers["first_name"] + " " + top_5_customers["last_name"]

)

display(top_5_customers[["customer_id", "full_name", "total_amount"]])

```

```python
# Determine the day with the highest average transaction value

average_transaction_by_day = (

    sales.groupby("day_of_week")["total_amount"]

    .mean()

    .sort_values(ascending=False)

)

print("Average transaction value by day:")

display(average_transaction_by_day)

print(

    "\nDay with the highest average transaction value:",

    average_transaction_by_day.idxmax()

)

```

## Discussion Questions

### 1. DateTime Benefits

Extracting the day of the week and month helps a retail business identify when customers shop most often. The business can use this information to plan staffing, inventory, promotions, and advertising around its busiest days and months.

### 2. GroupBy Value

Grouping data by categories, states, or time periods makes it easier to recognize patterns and compare performance. Instead of reviewing individual transactions, the business can quickly see which categories generate the most revenue, where most customers live, and when sales are strongest.

### 3. Business Applications

Based on the findings, the retail chain should consider increasing marketing and inventory for the Sports & Outdoors category because it generated the most revenue. The company could also run promotions during high-performing periods, focus on retaining its highest-spending customers, and examine why Thursday has the highest average transaction value.
